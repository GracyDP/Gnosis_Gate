```python
import pandas as pd
import numpy as np

def group_by_similarity(df, thresholds):
    """
    Raggruppa gli indici delle tuple simili per ciascun attributo.

    :param df: DataFrame con attributi numerici o stringhe
    :param thresholds: dict con soglie per ciascun attributo, es: {'A':1, 'B':1, 'C':1}
    :return: dizionario {colonna -> {id -> [id_simili]}}
    """
    result = {}

    for col in df.columns:
        threshold = thresholds[col]
        col_result = {}

        values = df[col].values

        for i in range(len(values)):
            simili = []
            for j in range(len(values)):
                val_i, val_j = values[i], values[j]
                if isinstance(val_i, str) or isinstance(val_j, str):
                    similarity = val_i == val_j
                else:
                    similarity = abs(val_i - val_j) <= threshold

                if similarity:
                    simili.append(j)
            col_result[i] = simili
        result[col] = col_result

    return result
```

```python
data = {
    'A': [175, 175, 176, 178, 169, 170, 170, 175],
    'B': [70, 75, 71, 81, 73, 62, 62, 69],
    'C': [40, 39, 40, 41, 37, 39, 39, 40]
}
df = pd.DataFrame(data)
thresholds = {'A': 1, 'B': 1, 'C': 1}

df
```

```python
similar_groups = group_by_similarity(df, thresholds)

for attr, group in similar_groups.items():
    print(f"{attr}: {group}")
```
